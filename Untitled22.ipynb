```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

%matplotlib inline
```

```python
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
```

```python
print(train.shape,test.shape)
print(train.head())
print()
print()
print(test.head())
```

```python
print(train.isnull().sum().sort_values(ascending=False).head(15))
print(train["SalePrice"].describe())
```

```python
Y = np.log1p(train["SalePrice"])
X = train.drop(["SalePrice","Id"], axis = 1)
test_ids = test["Id"]
X_test = test.drop("Id",axis = 1)
```

```python
all_data = pd.concat([X,X_test], keys=["train","test"])
num_cols = all_data.select_dtypes(include="number").columns
cat_cols =  all_data.select_dtypes(exclude="number").columns

all_data[num_cols] = all_data[num_cols].fillna(all_data[num_cols].median())
all_data[cat_cols] = all_data[cat_cols].fillna("None")
all_data = pd.get_dummies(all_data, columns=cat_cols)

X = all_data.loc["train"]
X_test = all_data.loc["test"]
```

```python
print(X.shape)
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)

print(x_train.shape)
print(x_test.shape)
```

```python
rf_model = RandomForestRegressor(
    n_estimators=350,
    random_state = 42,
    n_jobs=-1
)

rf_model.fit(x_train, y_train)

print("model training completed!")
```

```python
y_pred = rf_model.predict(x_test)
print("actual values:")
print(y_test)
print()
print("predicted value")
print(y_pred)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

y_pred = rf_model.predict(x_test)

rmse = np.sqrt(mean_squared_error(y_test,y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("RMSE (log):", rmse)
print("MAE (log):", mae)
print("R2 score:", r2)
```

```python
final_test = all_data.loc["test"]
```

```python
final_pred = np.expm1(rf_model.predict(final_test))
```

```python
submission = pd.DataFrame({"Id": test_ids, "SalePrice": final_pred})
print(submission)
```
